```python
import pandas as pd
import numpy as np

df = pd.read_csv('best_data.csv')

# FIX: Get the mean as a float, not a Series
y_pred_Averagemodel = df['Percent_Bleaching'].mean().round(6)

def RMSE(df, y_pred):
    y_true = df['Percent_Bleaching'].values
    mse = np.mean((y_true - y_pred)**2)
    return np.sqrt(mse)

def R2_score(df, y_pred):
    y_true = df['Percent_Bleaching'].values
    y_mean = np.mean(y_true)
    
    ss_res = np.sum((y_true - y_pred)**2)
    ss_tot = np.sum((y_true - y_mean)**2)
    
    r2 = 1 - (ss_res / ss_tot)
    
    print(f"SS_res : {ss_res:.6f}")
    print(f"SS_tot : {ss_tot:.6f}")
    
    return r2

print(f"RMSE : {RMSE(df, y_pred_Averagemodel):.6f}")
print("-" * 50)
print(f"R2_score : {R2_score(df, y_pred_Averagemodel):.6f}")
```
